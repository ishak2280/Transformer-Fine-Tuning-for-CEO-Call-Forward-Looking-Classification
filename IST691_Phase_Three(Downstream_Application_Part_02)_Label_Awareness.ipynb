```python
# ====================================================================================================
# IST 691 — Phase 3 (Application Module) | Label-Aware Semantic Drift & Interpretability (Fine-tuned)
# ====================================================================================================
# Purpose (APPLICATION, not evaluation):
# This module uses the *human/soft labels* in the original dataset as an interpretability layer to
# study how labeled themes (e.g., strategic_focus = "Artificial Intelligence Focus") evolve over time.
#
# It produces:
#   A) Quarter-level label prevalence trends (ABSOLUTE counts + RELATIVE shares)
#   B) Label-conditioned semantic subtopic dynamics (within the label: cluster composition over time)
#   C) Representative exemplar sentences (per quarter + per dominant cluster) to support interpretation
#
# Notes:
# - This module DOES NOT re-train or re-embed anything.
# - It consumes existing Phase 2 Part 3 cluster assignments (KMeans K=30) + the original labeled CSV.
# - All key artifacts are BOTH (1) saved locally and (2) previewed/rendered directly in the notebook.
# ====================================================================================================

from __future__ import annotations

import os
import sys
import json
import time
import hashlib
from dataclasses import dataclass
from pathlib import Path
from datetime import datetime
from typing import Dict, List, Tuple, Optional

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# -----------------------------
# CONFIG
# -----------------------------
CFG: Dict[str, object] = {
    # Root resolution
    "project_root_env": "IST691_PROJECT_ROOT",
    "phase2_root_dirname": "IST691_phase2_output",
    "phase3_root_dirname": "IST691_phase3_output",

    # Phase 2 Part 3 (cluster assignments) — auto-discover newest matching folder OR hard pin
    "part3_dir_hint": "part3_compare_v2__no_pca__kmeansK30",   # used for auto-selection
    "part3_dir_override": None,  # set to Path(...) string to hard-pin if desired

    # Required Phase 2 files (KMeans K=30 assignments)
    "baseline_assignments_name": "baseline_kmeans_k30__assignments.csv",
    "finetuned_assignments_name": "finetuned_kmeans_k30__assignments.csv",

    # Label CSV (original dataset)
    "label_csv_name": "belief_extraction_results_deduped_year_quarter_flattened_repeat_deduped.csv",
    "label_csv_override": None,  # set to Path(...) string to hard-pin if desired

    # Column schema
    "assign_id_col": "sentence_id",         # in assignments
    "assign_quarter_col": "year_quarter",   # in assignments
    "assign_text_col": "belief_sentence",   # in assignments
    "assign_cluster_col": "cluster_id",     # in assignments

    "label_id_col": "hash_id",              # in label CSV
    "label_cols": ["strategic_focus", "temporal_framing", "certainty_level", "market_position"],

    # Label-aware application scope
    "labelaware_cols": ["strategic_focus", "temporal_framing", "market_position"],  # keep compact
    "labelaware_top_values": 8,                 # top label values displayed; remainder -> "Other"
    "labelaware_samples_per_quarter": 6,        # exemplars by quarter
    "labelaware_samples_per_cluster": 4,        # exemplars by (quarter, dominant cluster) within focus label
    "samples_seed": 42,

    # Focus label (for label-conditioned subtopic drift)
    "focus_label_col": "strategic_focus",
    "focus_label_value": "Artificial Intelligence Focus",

    # Time aggregation (quarter-level)
    # Optional: also export a year-level view derived from quarters
    "also_export_year_view": True,

    # Output naming
    "run_tag": "phase3_application__labelaware_semantic_drift__finetuned_only",
}


# -----------------------------
# Utilities
# -----------------------------
def _md5_file(path: Path, chunk_size: int = 1024 * 1024) -> str:
    h = hashlib.md5()
    with path.open("rb") as f:
        while True:
            b = f.read(chunk_size)
            if not b:
                break
            h.update(b)
    return h.hexdigest()


def _now_tag() -> str:
    return datetime.now().strftime("%Y%m%d_%H%M%S")


def _safe_mkdir(p: Path) -> None:
    p.mkdir(parents=True, exist_ok=True)


def _print_header(title: str) -> None:
    line = "=" * 108
    print("\n" + line)
    print(title)
    print(line)


def _section(title: str) -> None:
    print("\n" + "-" * 92)
    print(f"[ {title} ]")
    print("-" * 92)


def _preview_df(df: pd.DataFrame, title: str, n: int = 10) -> None:
    _section(f"preview: {title} (top {n})")
    if df is None or df.empty:
        print("(empty)")
        return
    with pd.option_context("display.max_colwidth", 120, "display.width", 160):
        print(df.head(n).to_string(index=False))


def _save_csv(df: pd.DataFrame, path: Path, preview_title: str, preview_n: int = 10) -> None:
    df.to_csv(path, index=False)
    print(f"Saved CSV: {path}")
    _preview_df(df, preview_title, n=preview_n)


def _save_show_fig(fig: plt.Figure, out_path: Path) -> None:
    fig.savefig(out_path, dpi=220, bbox_inches="tight")
    print(f"Saved figure: {out_path}")
    plt.show()
    plt.close(fig)


def _quarter_sort_key(q: str) -> Tuple[int, int]:
    # expects "YYYY-Qn"
    try:
        y, qn = q.split("-Q")
        return int(y), int(qn)
    except Exception:
        return (9999, 9)


def _quarter_order(quarters: List[str]) -> List[str]:
    return sorted(list(set(quarters)), key=_quarter_sort_key)


def _year_from_quarter(q: str) -> Optional[int]:
    try:
        return int(q.split("-Q")[0])
    except Exception:
        return None


def _as_year_table(quarter_table: pd.DataFrame, time_col: str = "year_quarter") -> pd.DataFrame:
    # Converts a quarter-level table with columns: [time_col, ..., "n"] into a year-level aggregation.
    d = quarter_table.copy()
    d["year"] = d[time_col].map(_year_from_quarter)
    d = d.dropna(subset=["year"])
    d["year"] = d["year"].astype(int)
    return d


def _auto_find_project_root() -> Path:
    # Priority:
    # 1) ENV var
    # 2) cwd containing phase2 folder
    env = os.getenv(str(CFG["project_root_env"]), "").strip()
    if env:
        p = Path(env).expanduser().resolve()
        if p.exists():
            return p

    cwd = Path.cwd().resolve()
    if (cwd / str(CFG["phase2_root_dirname"])).exists():
        return cwd

    # fallback: walk up a few levels
    cur = cwd
    for _ in range(6):
        if (cur / str(CFG["phase2_root_dirname"])).exists():
            return cur
        if cur.parent == cur:
            break
        cur = cur.parent

    # final fallback: cwd
    return cwd


def _pick_part3_dir(phase2_root: Path) -> Path:
    override = CFG.get("part3_dir_override")
    if override:
        p = Path(str(override)).expanduser().resolve()
        if not p.exists():
            raise FileNotFoundError(f"part3_dir_override not found: {p}")
        return p

    hint = str(CFG["part3_dir_hint"])
    candidates = [p for p in phase2_root.iterdir() if p.is_dir() and hint in p.name]
    if not candidates:
        # fallback: any dir containing "part3_compare"
        candidates = [p for p in phase2_root.iterdir() if p.is_dir() and "part3_compare" in p.name]
    if not candidates:
        raise FileNotFoundError(f"No Phase 2 Part 3 folder found under: {phase2_root}")

    # newest by mtime
    candidates.sort(key=lambda p: p.stat().st_mtime, reverse=True)
    return candidates[0]


def _resolve_label_csv(project_root: Path) -> Path:
    override = CFG.get("label_csv_override")
    if override:
        p = Path(str(override)).expanduser().resolve()
        if not p.exists():
            raise FileNotFoundError(f"label_csv_override not found: {p}")
        return p

    p = (project_root / str(CFG["label_csv_name"])).resolve()
    if not p.exists():
        raise FileNotFoundError(f"Label CSV not found at expected location: {p}")
    return p


# -----------------------------
# Core computations
# -----------------------------
def build_time_label_table(df_merged: pd.DataFrame, time_col: str, label_col: str) -> pd.DataFrame:
    d = df_merged[[time_col, label_col]].dropna().copy()
    g = d.groupby([time_col, label_col]).size().reset_index(name="n")
    g["time_total"] = g.groupby(time_col)["n"].transform("sum")
    g["share"] = g["n"] / g["time_total"]
    g = g.sort_values([time_col, "share"], ascending=[True, False])
    return g


def pivot_topk_with_other(long_tbl: pd.DataFrame, time_col: str, label_col: str, top_k: int) -> Tuple[pd.DataFrame, pd.DataFrame]:
    # wide counts and wide shares with "Other"
    quarters = _quarter_order(long_tbl[time_col].unique().tolist())
    counts = long_tbl.pivot(index=time_col, columns=label_col, values="n").fillna(0).reindex(quarters)

    top = counts.sum(axis=0).sort_values(ascending=False).head(top_k).index.tolist()
    other_cols = [c for c in counts.columns if c not in top]

    out_counts = counts[top].copy()
    out_counts["Other"] = counts[other_cols].sum(axis=1) if other_cols else 0

    denom = out_counts.sum(axis=1).replace(0, np.nan)
    out_share = (out_counts.T / denom).T.fillna(0)
    return out_counts, out_share


def plot_stacked_share(share_wide: pd.DataFrame, title: str, ylabel: str, out_path: Path) -> None:
    fig = plt.figure(figsize=(14, 6))
    ax = plt.gca()
    x = np.arange(len(share_wide.index))
    bottoms = np.zeros(len(x), dtype=float)

    cols = list(share_wide.columns)
    if "Other" in cols:
        cols = ["Other"] + [c for c in cols if c != "Other"]

    for col in cols:
        ax.bar(x, share_wide[col].to_numpy(dtype=float), bottom=bottoms, label=str(col), width=0.78)
        bottoms += share_wide[col].to_numpy(dtype=float)

    ax.set_title(title, fontsize=14, fontweight="bold")
    ax.set_xlabel("Quarter")
    ax.set_ylabel(ylabel)
    ax.set_xticks(x)
    ax.set_xticklabels(share_wide.index.tolist(), rotation=45, ha="right")
    ax.grid(axis="y", alpha=0.25)
    ax.legend(loc="center left", bbox_to_anchor=(1.01, 0.5), fontsize=8, title="Label", title_fontsize=9)
    _save_show_fig(fig, out_path)


def plot_stacked_counts(counts_wide: pd.DataFrame, title: str, ylabel: str, out_path: Path) -> None:
    fig = plt.figure(figsize=(14, 6))
    ax = plt.gca()
    x = np.arange(len(counts_wide.index))
    bottoms = np.zeros(len(x), dtype=float)

    cols = list(counts_wide.columns)
    if "Other" in cols:
        cols = ["Other"] + [c for c in cols if c != "Other"]

    for col in cols:
        ax.bar(x, counts_wide[col].to_numpy(dtype=float), bottom=bottoms, label=str(col), width=0.78)
        bottoms += counts_wide[col].to_numpy(dtype=float)

    ax.set_title(title, fontsize=14, fontweight="bold")
    ax.set_xlabel("Quarter")
    ax.set_ylabel(ylabel)
    ax.set_xticks(x)
    ax.set_xticklabels(counts_wide.index.tolist(), rotation=45, ha="right")
    ax.grid(axis="y", alpha=0.25)
    ax.legend(loc="center left", bbox_to_anchor=(1.01, 0.5), fontsize=8, title="Label", title_fontsize=9)
    _save_show_fig(fig, out_path)


def focus_label_cluster_drift(df_merged: pd.DataFrame, time_col: str, label_col: str, label_val: str, cluster_col: str) -> pd.DataFrame:
    d = df_merged[df_merged[label_col] == label_val].copy()
    d = d.dropna(subset=[time_col, cluster_col])
    g = d.groupby([time_col, cluster_col]).size().reset_index(name="n")
    g["time_total_within_label"] = g.groupby(time_col)["n"].transform("sum")
    g["share_within_label"] = g["n"] / g["time_total_within_label"]
    g = g.sort_values([time_col, "n"], ascending=[True, False])
    return g


def plot_focus_label_top_clusters_share(drift: pd.DataFrame, time_col: str, cluster_col: str, top_k: int, title: str, out_path: Path) -> pd.DataFrame:
    # Determine top clusters globally within the focus label
    clusters = drift.groupby(cluster_col)["n"].sum().sort_values(ascending=False)
    top = clusters.head(top_k).index.tolist()

    quarters = _quarter_order(drift[time_col].unique().tolist())
    wide = drift[drift[cluster_col].isin(top)].pivot(index=time_col, columns=cluster_col, values="share_within_label").fillna(0).reindex(quarters)
    wide["Other"] = 1.0 - wide.sum(axis=1)
    wide["Other"] = wide["Other"].clip(lower=0.0)

    plot_stacked_share(
        wide,
        title=title,
        ylabel="Share within label",
        out_path=out_path
    )
    return wide


def plot_focus_label_top_clusters_counts(drift: pd.DataFrame, time_col: str, cluster_col: str, top_k: int, title: str, out_path: Path) -> pd.DataFrame:
    clusters = drift.groupby(cluster_col)["n"].sum().sort_values(ascending=False)
    top = clusters.head(top_k).index.tolist()

    quarters = _quarter_order(drift[time_col].unique().tolist())
    wide = drift[drift[cluster_col].isin(top)].pivot(index=time_col, columns=cluster_col, values="n").fillna(0).reindex(quarters)
    wide["Other"] = drift[~drift[cluster_col].isin(top)].groupby(time_col)["n"].sum().reindex(quarters).fillna(0)

    plot_stacked_counts(
        wide,
        title=title,
        ylabel="Number of sentences",
        out_path=out_path
    )
    return wide


def sample_exemplars_by_quarter(df_merged: pd.DataFrame, time_col: str, text_col: str, id_col: str, label_col: str, label_val: str,
                               n_per_quarter: int, seed: int) -> pd.DataFrame:
    rng = np.random.default_rng(seed)
    d = df_merged[df_merged[label_col] == label_val].dropna(subset=[time_col, text_col]).copy()
    rows = []
    for q in _quarter_order(d[time_col].unique().tolist()):
        part = d[d[time_col] == q]
        if part.empty:
            continue
        take_n = min(n_per_quarter, len(part))
        idx = rng.choice(part.index.to_numpy(), size=take_n, replace=False)
        take = part.loc[idx, [id_col, time_col, label_col, text_col]].copy()
        take["sample_rank_in_quarter"] = np.arange(1, len(take) + 1)
        rows.append(take)
    return pd.concat(rows, ignore_index=True) if rows else pd.DataFrame()


def sample_exemplars_by_quarter_and_cluster(df_merged: pd.DataFrame, time_col: str, text_col: str, id_col: str,
                                           label_col: str, label_val: str, cluster_col: str,
                                           top_clusters_per_quarter: int, samples_per_cluster: int, seed: int) -> pd.DataFrame:
    rng = np.random.default_rng(seed)
    d = df_merged[df_merged[label_col] == label_val].dropna(subset=[time_col, text_col, cluster_col]).copy()

    out_rows = []
    for q in _quarter_order(d[time_col].unique().tolist()):
        part = d[d[time_col] == q]
        if part.empty:
            continue
        # dominant clusters within this quarter (inside label)
        topc = part.groupby(cluster_col).size().sort_values(ascending=False).head(top_clusters_per_quarter).index.tolist()
        for c in topc:
            pc = part[part[cluster_col] == c]
            take_n = min(samples_per_cluster, len(pc))
            idx = rng.choice(pc.index.to_numpy(), size=take_n, replace=False)
            take = pc.loc[idx, [id_col, time_col, cluster_col, label_col, text_col]].copy()
            take["cluster_size_within_quarter_label"] = int(len(pc))
            take["sample_rank_in_cluster"] = np.arange(1, len(take) + 1)
            out_rows.append(take)

    return pd.concat(out_rows, ignore_index=True) if out_rows else pd.DataFrame()


# -----------------------------
# Main
# -----------------------------
def main() -> None:
    t0 = time.time()
    project_root = _auto_find_project_root()
    phase2_root = (project_root / str(CFG["phase2_root_dirname"])).resolve()
    if not phase2_root.exists():
        raise FileNotFoundError(f"Phase 2 root not found: {phase2_root}")

    part3_dir = _pick_part3_dir(phase2_root)
    label_csv = _resolve_label_csv(project_root)

    # Output dirs (no overwrite: new timestamped run folder)
    phase3_root = (project_root / str(CFG["phase3_root_dirname"])).resolve()
    _safe_mkdir(phase3_root)
    run_dir = phase3_root / f"{CFG['run_tag']}__{_now_tag()}"
    figures_dir = run_dir / "figures"
    reports_dir = run_dir / "reports"
    _safe_mkdir(run_dir)
    _safe_mkdir(figures_dir)
    _safe_mkdir(reports_dir)

    _print_header("IST 691 — Phase 3 | Application: Label-Aware Semantic Drift & Interpretability (Fine-tuned)")

    _section("startup")
    print("timestamp   :", datetime.now().strftime("%Y-%m-%d %H:%M:%S"))
    print("platform    :", sys.platform)
    print("python      :", sys.version.split()[0])
    print("cwd         :", str(Path.cwd().resolve()))
    print("project_root:", str(project_root))
    print("phase2_root :", str(phase2_root))
    print("part3_dir   :", str(part3_dir))
    print("phase3_run  :", str(run_dir))

    # Load Phase 2 Part 3 KMeans assignments (fine-tuned only for this application)
    finetuned_path = (part3_dir / str(CFG["finetuned_assignments_name"])).resolve()
    if not finetuned_path.exists():
        raise FileNotFoundError(f"Missing fine-tuned assignments: {finetuned_path}")

    _section("inputs: Phase 2 Part 3 (fine-tuned) assignments")
    print("finetuned assignments:", str(finetuned_path))
    print("finetuned_md5        :", _md5_file(finetuned_path))
    df_fine = pd.read_csv(finetuned_path)
    _preview_df(df_fine, "fine-tuned assignments", n=8)

    # Validate assignment schema
    id_col = str(CFG["assign_id_col"])
    quarter_col = str(CFG["assign_quarter_col"])
    text_col = str(CFG["assign_text_col"])
    cluster_col = str(CFG["assign_cluster_col"])

    missing = [c for c in [id_col, quarter_col, text_col, cluster_col] if c not in df_fine.columns]
    if missing:
        raise KeyError(f"Missing required columns in assignments: {missing}. Found: {list(df_fine.columns)[:40]}")

    df_fine[id_col] = df_fine[id_col].astype(str)

    # Load label CSV (original dataset)
    _section("inputs: label table (original dataset; safe merge)")
    print("label_csv:", str(label_csv))
    print("label_md5:", _md5_file(label_csv))
    df_lab = pd.read_csv(label_csv)

    label_id = str(CFG["label_id_col"])
    if label_id not in df_lab.columns:
        raise KeyError(f"label_id_col '{label_id}' not found in label CSV. Columns: {list(df_lab.columns)[:40]}")

    # Only import label columns (avoid year/text collisions)
    label_cols = list(CFG["label_cols"])
    for lc in label_cols:
        if lc not in df_lab.columns:
            raise KeyError(f"Label column missing in label CSV: {lc}")

    df_lab = df_lab[[label_id] + label_cols].copy()
    df_lab[label_id] = df_lab[label_id].astype(str)

    # Merge (assignment id aligns with label hash_id)
    merged_fine = df_fine.merge(df_lab, left_on=id_col, right_on=label_id, how="left")
    merged_fine = merged_fine.drop(columns=[label_id])

    coverage = float(merged_fine[label_cols].notna().all(axis=1).mean())
    print(f"label coverage (all label cols non-null): {coverage:.4f}")
    _preview_df(merged_fine[[id_col, quarter_col, cluster_col] + label_cols].head(8), "merged fine schema", n=8)

    # Save merged snapshot (light)
    merged_snapshot = merged_fine[[id_col, quarter_col, cluster_col] + label_cols].copy()
    _save_csv(merged_snapshot, reports_dir / "finetuned_kmeans_k30__merged_labels_snapshot.csv",
              "finetuned_kmeans_k30__merged_labels_snapshot.csv", preview_n=10)

    # ---------------------------------------
    # A) Label prevalence trends over time
    # ---------------------------------------
    _section("application A: label prevalence over time (quarter) — counts + shares")
    labelaware_cols = list(CFG["labelaware_cols"])
    top_values = int(CFG["labelaware_top_values"])

    prevalence_artifacts: List[Path] = []

    for lab in labelaware_cols:
        tl = build_time_label_table(merged_fine, quarter_col, lab)

        out_long = reports_dir / f"finetuned_quarter_label_distribution__{lab}.csv"
        _save_csv(tl, out_long, f"quarter × {lab} (long)", preview_n=12)
        prevalence_artifacts.append(out_long)

        counts_w, share_w = pivot_topk_with_other(tl, quarter_col, lab, top_values)

        # Export wide views (easier for report / inspection)
        out_counts = reports_dir / f"finetuned_quarter_label_distribution__{lab}__counts_wide.csv"
        out_share = reports_dir / f"finetuned_quarter_label_distribution__{lab}__share_wide.csv"
        _save_csv(counts_w.reset_index().rename(columns={quarter_col: "year_quarter"}), out_counts, f"{lab} counts wide", preview_n=10)
        _save_csv(share_w.reset_index().rename(columns={quarter_col: "year_quarter"}), out_share, f"{lab} share wide", preview_n=10)

        # Plot: stacked share
        plot_stacked_share(
            share_w,
            title=f"Fine-tuned — {lab} Over Time (Quarter) [Share]",
            ylabel="Share of sentences",
            out_path=figures_dir / f"figA_finetuned_{lab}_quarter_share.png"
        )

        # Plot: stacked absolute counts
        plot_stacked_counts(
            counts_w,
            title=f"Fine-tuned — {lab} Over Time (Quarter) [Count]",
            ylabel="Number of sentences",
            out_path=figures_dir / f"figA_finetuned_{lab}_quarter_count.png"
        )

        # Optional: year-level aggregation
        if bool(CFG["also_export_year_view"]):
            y = _as_year_table(tl, time_col=quarter_col)
            yg = y.groupby(["year", lab], as_index=False)["n"].sum()
            yg["year_total"] = yg.groupby("year")["n"].transform("sum")
            yg["share"] = yg["n"] / yg["year_total"]
            yg = yg.sort_values(["year", "share"], ascending=[True, False])

            out_ylong = reports_dir / f"finetuned_year_label_distribution__{lab}.csv"
            _save_csv(yg, out_ylong, f"year × {lab} (long)", preview_n=12)

            # wide for year
            counts_y = yg.pivot(index="year", columns=lab, values="n").fillna(0)
            top = counts_y.sum(axis=0).sort_values(ascending=False).head(top_values).index.tolist()
            other_cols = [c for c in counts_y.columns if c not in top]
            counts_yw = counts_y[top].copy()
            counts_yw["Other"] = counts_y[other_cols].sum(axis=1) if other_cols else 0
            share_yw = (counts_yw.T / counts_yw.sum(axis=1).replace(0, np.nan)).T.fillna(0)

            plot_stacked_share(
                share_yw,
                title=f"Fine-tuned — {lab} Over Time (Year) [Share]",
                ylabel="Share of sentences",
                out_path=figures_dir / f"figA_finetuned_{lab}_year_share.png"
            )
            plot_stacked_counts(
                counts_yw,
                title=f"Fine-tuned — {lab} Over Time (Year) [Count]",
                ylabel="Number of sentences",
                out_path=figures_dir / f"figA_finetuned_{lab}_year_count.png"
            )

    # ---------------------------------------------------------
    # B) Label-conditioned subtopic dynamics (within focus label)
    # ---------------------------------------------------------
    _section("application B: focus label → cluster composition over time (quarter) — counts + shares")
    focus_lab = str(CFG["focus_label_col"])
    focus_val = str(CFG["focus_label_value"])

    if focus_lab not in merged_fine.columns:
        raise KeyError(f"focus_label_col '{focus_lab}' not present in merged data.")
    drift = focus_label_cluster_drift(merged_fine, quarter_col, focus_lab, focus_val, cluster_col)

    out_drift = reports_dir / f"finetuned_focuslabel_cluster_drift__{focus_lab}__{focus_val.replace(' ', '_')}.csv"
    _save_csv(drift, out_drift, f"focus label drift: {focus_lab}={focus_val}", preview_n=15)

    # Plot top clusters within the focus label (share + count)
    top_clusters_k = 12
    wide_share = plot_focus_label_top_clusters_share(
        drift, quarter_col, cluster_col, top_k=top_clusters_k,
        title=f"Fine-tuned — Within '{focus_val}': Cluster Composition Over Time (Quarter) [Share]",
        out_path=figures_dir / f"figB_focuslabel_clusters_quarter_share__{focus_val.replace(' ', '_')}.png"
    )
    _preview_df(wide_share.reset_index().head(10), "focus label cluster share (wide)", n=10)

    wide_counts = plot_focus_label_top_clusters_counts(
        drift, quarter_col, cluster_col, top_k=top_clusters_k,
        title=f"Fine-tuned — Within '{focus_val}': Cluster Composition Over Time (Quarter) [Count]",
        out_path=figures_dir / f"figB_focuslabel_clusters_quarter_count__{focus_val.replace(' ', '_')}.png"
    )
    _preview_df(wide_counts.reset_index().head(10), "focus label cluster count (wide)", n=10)

    # ---------------------------------------------------------
    # C) Exemplars for interpretability (sentences)
    # ---------------------------------------------------------
    _section("application C: exemplar sentences (quarter + cluster) for interpretability")
    exemplars_q = sample_exemplars_by_quarter(
        merged_fine, quarter_col, text_col, id_col,
        label_col=focus_lab, label_val=focus_val,
        n_per_quarter=int(CFG["labelaware_samples_per_quarter"]),
        seed=int(CFG["samples_seed"]),
    )
    out_ex_q = reports_dir / f"finetuned_exemplars_by_quarter__{focus_val.replace(' ', '_')}.csv"
    _save_csv(exemplars_q, out_ex_q, "exemplars by quarter (focus label)", preview_n=25)

    exemplars_qc = sample_exemplars_by_quarter_and_cluster(
        merged_fine, quarter_col, text_col, id_col,
        label_col=focus_lab, label_val=focus_val,
        cluster_col=cluster_col,
        top_clusters_per_quarter=2,
        samples_per_cluster=int(CFG["labelaware_samples_per_cluster"]),
        seed=int(CFG["samples_seed"]) + 1,
    )
    out_ex_qc = reports_dir / f"finetuned_exemplars_by_quarter_and_cluster__{focus_val.replace(' ', '_')}.csv"
    _save_csv(exemplars_qc, out_ex_qc, "exemplars by quarter+cluster (focus label)", preview_n=30)

    # ---------------------------------------------------------
    # Compact summary tables (console + export)
    # ---------------------------------------------------------
    _section("summary tables (compact)")
    # Summary A: overall label distributions (top values) for each label column
    summary_rows = []
    for lab in labelaware_cols:
        vc = merged_fine[lab].value_counts(dropna=True)
        top = vc.head(10)
        tot = int(vc.sum())
        for k, v in top.items():
            summary_rows.append({"label_col": lab, "label_value": str(k), "n": int(v), "share": float(v / tot)})
    summary_df = pd.DataFrame(summary_rows).sort_values(["label_col", "n"], ascending=[True, False])
    _save_csv(summary_df, reports_dir / "summary_top_label_values.csv", "summary_top_label_values.csv", preview_n=25)

    # Summary B: focus label total volume by quarter
    focus_q = merged_fine[merged_fine[focus_lab] == focus_val].groupby(quarter_col).size().reset_index(name="n")
    focus_q = focus_q.sort_values(quarter_col, key=lambda s: s.map(_quarter_sort_key))
    focus_q["share_of_all_sentences"] = focus_q["n"] / merged_fine.groupby(quarter_col).size().reindex(focus_q[quarter_col]).to_numpy()
    _save_csv(focus_q, reports_dir / f"summary_focuslabel_volume_by_quarter__{focus_val.replace(' ', '_')}.csv",
              f"focus label volume by quarter: {focus_val}", preview_n=20)

    # Simple line plot: absolute focus label volume over time
    fig = plt.figure(figsize=(12, 4.5))
    ax = plt.gca()
    x = np.arange(len(focus_q))
    ax.plot(x, focus_q["n"].to_numpy(), marker="o")
    ax.set_title(f"Fine-tuned — '{focus_val}' Volume Over Time (Quarter) [Count]", fontsize=13, fontweight="bold")
    ax.set_xlabel("Quarter")
    ax.set_ylabel("Number of sentences")
    ax.set_xticks(x)
    ax.set_xticklabels(focus_q[quarter_col].tolist(), rotation=45, ha="right")
    ax.grid(alpha=0.25)
    _save_show_fig(fig, figures_dir / f"figC_focuslabel_volume_quarter_count__{focus_val.replace(' ', '_')}.png")

    # Manifest
    _section("export manifest")
    manifest = {
        "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "project_root": str(project_root),
        "phase2_root": str(phase2_root),
        "part3_dir": str(part3_dir),
        "label_csv": str(label_csv),
        "inputs": {
            "finetuned_assignments": str(finetuned_path),
        },
        "outputs": {
            "run_dir": str(run_dir),
            "figures_dir": str(figures_dir),
            "reports_dir": str(reports_dir),
        },
        "focus_label": {"col": focus_lab, "value": focus_val},
        "cfg": {k: (str(v) if isinstance(v, Path) else v) for k, v in CFG.items()},
    }
    manifest_path = run_dir / "phase3_labelaware_application_manifest.json"
    manifest_path.write_text(json.dumps(manifest, indent=2), encoding="utf-8")
    print("Saved manifest:", str(manifest_path))

    _print_header("Phase 3 label-aware application completed (artifacts saved + rendered in notebook)")
    print("run_dir   :", run_dir)
    print("reports   :", reports_dir)
    print("figures   :", figures_dir)
    print(f"elapsed_s : {time.time() - t0:.2f}")


# Execute
main()
```
